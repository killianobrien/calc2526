## Newton's method 

In this notebook we investigate the sequence of estimates
$$x_0, x_1, x_2, x_3, \dots $$
that come from the recurrence relation
$$x_0 = 3, \quad \forall n \geq 1 \, \, x_{n+1} = x_n - \frac{x_n^2 - 5}{2 x_n}.$$
This recurrence comes from Newton's method applied to the function defined by 
$$f(x) = x^2 -5,$$
and satisfies the convergence property
$$\lim_{n \to \infty} x_n = \sqrt{5},$$
that we proved in thiis week's tutorial. 

```python
import sympy
```

```python
x = sympy.Integer(3)
x
```

```python
type(x)
```

```python
ests = [x]
for n in range(10):
    x = ests[-1]
    y = x - (x**2 - 5)/(2*x)
    ests.append(y)
```

```python
len(ests)
```

```python
for e in ests:
    print(e)
    print('-----------')
```

```python
sympy.RealNumber(ests[-1]**2, 1000)
```

```python
for e in ests:
    y = sympy.RealNumber(e**2, 1000)
    print(y)
    print('-------------')
```
